# Equilibration Results

Energy minimization, NVT, and NPT diagnostics.

## 1. Energy minimization — `potential.xvg`

```python
import matplotlib.pyplot as plt
import numpy as np

# Load data, skipping comment lines starting with # or @
x, y = [], []
filename = "./potential.xvg"

with open(filename, "r") as f:
    for line in f:
        if line.startswith(("#", "@")):
            continue
        cols = line.split()
        if len(cols) >= 2:
            x.append(float(cols[0]))
            y.append(float(cols[1]))

plt.figure(figsize=(8, 5))

# Points + connecting line
plt.plot(
    x, y,
    marker="o",
    markersize=4,
    linewidth=1.0,
    color="blue"
)

plt.title("Energy minimization", fontsize=18)
plt.xlabel("EM step")
plt.ylabel("Potential energy (kJ/mol)")

# Scientific notation on y axis
plt.ticklabel_format(
    axis="y",
    style="sci",
    scilimits=(0, 0),
    useMathText=True
)

plt.xlabel("EM step", fontsize=16)
plt.ylabel("Potential energy (kJ/mol)", fontsize=16)
plt.xticks(np.arange(min(x), max(x) + 1, 10), fontsize=14)
plt.yticks(fontsize=14)
plt.tick_params(axis="both", which="major", length=6, width=1.2)

plt.tight_layout()
plt.savefig("potential_ene.png", dpi=600)
plt.show()
```

## 2. Temperature (NVT) — `temperature.xvg`

```python
x, y = [], []
filename = "./temperature.xvg"

with open(filename, "r") as f:
    for line in f:
        if line.startswith(("#", "@")):
            continue
        cols = line.split()
        if len(cols) >= 2:
            x.append(float(cols[0]))
            y.append(float(cols[1]))

avg = np.mean(y)
std = np.std(y)

plt.figure(figsize=(8, 5))
plt.plot(np.array(x)/1000, y, color="blue", linestyle="-", marker="o", markersize=4, linewidth=0.2)
plt.axhline(avg, color="red", linestyle="-", linewidth=1.5, label=f"average = {avg:.2f} \u00b1 {std:.2f}")
plt.legend(fontsize=14)
plt.title("NVT equilibration", fontsize=18)
plt.xlabel("Time (ns)", fontsize=16)
plt.ylabel("Temperature (K)", fontsize=16)
plt.tick_params(axis="both", which="major", length=6, width=1.2)
plt.xlim(-0.05,1.05)
plt.ylim(275,328)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)

plt.tight_layout()
plt.savefig('T_equilib.png', dpi=600)
plt.show()
```

## 3. Pressure (NPT) — `pressure.xvg`

```python
x, y = [], []
filename = "./pressure.xvg"

with open(filename, "r") as f:
    for line in f:
        if line.startswith(("#", "@")):
            continue
        cols = line.split()
        if len(cols) >= 2:
            x.append(float(cols[0]))
            y.append(float(cols[1]))

avg = np.mean(y)
std = np.std(y)

plt.figure(figsize=(8, 5))
plt.plot(np.array(x[::1])/1000, y[::1], color="gold", linestyle="none", marker="o", markersize=2)
plt.axhline(avg, color="red", linestyle="-", linewidth=1.5, label=f"average = {avg:.2f} \u00b1 {std:.2f}")
plt.legend(fontsize=14)
plt.title("Pressure equilibration", fontsize=18)
plt.xlabel("Time (ns)", fontsize=16)
plt.ylabel("Pressure (bar)", fontsize=16)
plt.tick_params(axis="both", which="major", length=6, width=1.2)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)

plt.tight_layout()
plt.savefig('P_equilib.png', dpi=600)
plt.show()
```

## 4. Density (NPT) — `density.xvg`

```python
x, y = [], []
filename = "density.xvg"

with open(filename, "r") as f:
    for line in f:
        if line.startswith(("#", "@")):
            continue
        cols = line.split()
        if len(cols) >= 2:
            x.append(float(cols[0]))
            y.append(float(cols[1]))

avg = np.mean(y)
std = np.std(y)

plt.figure(figsize=(8, 5))
plt.plot(np.array(x[::1])/1000, y[::1], color="green", linestyle="none", marker="o", markersize=2)
plt.axhline(avg, color="red", linestyle="-", linewidth=1.5, label=f"average = {avg:.2f} \u00b1 {std:.2f}")
plt.legend(fontsize=14)
plt.title("Density equilibration", fontsize=18)
plt.xlabel("Time (ns)", fontsize=16)
plt.ylabel(r"Density (kg/m$^3$)", fontsize=16)
plt.tick_params(axis="both", which="major", length=6, width=1.2)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)

plt.tight_layout()
plt.savefig("Density_equilib.png", dpi=600)
plt.show()
```
